````python
import mitsuba_config  # active la variante Mitsuba configuree pour cette machine
from scene_generation import make_simple_scene, render_scene
import mitsuba as mi
````

````python
scene_dict = make_simple_scene(sphere_radius=1.0, resolution=256)
image = render_scene(scene_dict)
````

````python
import matplotlib.pyplot as plt
import numpy as np

img_np = np.array(image)
plt.figure(figsize=(5, 5))
plt.imshow(np.clip(img_np ** (1 / 2.2), 0, 1))  # correction gamma pour affichage
plt.axis("off")
plt.title("Rendu Mitsuba 3")
plt.show()
````

## Variante : comparer plusieurs rayons de sphere

````python
radii = [0.5, 1.0, 1.5]
fig, axes = plt.subplots(1, len(radii), figsize=(5 * len(radii), 5))

for ax, r in zip(axes, radii):
    scene_dict = make_simple_scene(sphere_radius=r, resolution=256)
    img = np.array(render_scene(scene_dict))
    ax.imshow(np.clip(img ** (1 / 2.2), 0, 1))
    ax.set_title(f"radius = {r}")
    ax.axis("off")

plt.tight_layout()
plt.show()
````
